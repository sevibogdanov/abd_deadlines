```python
import pandas as pd
import datetime
import mcfg_abd

#кс

conn = mcfg_abd.conn2()
cur = conn.cursor()
cur.execute(f"""delete from deadline_ks;""")
conn.commit()
conn.close()

def to_time(txt):
    mon_dict = {'сентября':9,'октября':10,'ноября':11,'декабря':12}
    txt = txt.split(' ')
    day = int(txt[0])
    month = mon_dict[txt[1]]
    year = 2024
    hour = 19
    return datetime.datetime(year,month,day,hour)

df = pd.read_html(r"C:\Users\Admin\Desktop\abd_bot\tables.html",encoding='utf8')
df = df[0][['Дедлайн','Название']]
df = df[~df['Дедлайн'].isna()].rename(columns={'Дедлайн':'dtime','Название':'d_nm'})
df['d_type'] = df['dtime'].apply(lambda x: 'кс')
df['dtime'] = df['dtime'].apply(to_time)
df['link'] = df['dtime'].apply(lambda x: 'https://lab.karpov.courses/learning/609/schedule/')
df.to_sql('deadline_ks',con=mcfg_abd.my_connection(),index=False,if_exists='append')

#gs

# import pandas as pd
# import datetime
# import mcfg_abd

# df = pd.read_csv('https://docs.google.com/spreadsheets/d/e/2PACX-1vR9qLWoMIogTOWiCW4_-R8OETkR667_jyEYj57e8Auh6VkqQHfQbSpM_4-cBSPoegJcliP_SP4AcoJ6/pub?gid=1036789569&single=true&output=csv', encoding = 'utf8')
# df['dtime']  = pd.to_datetime(df['Срок сдачи'] +' '+ df['Время сдачи'],format="%d.%m.%Y %H:%M")
# df.rename(columns={
#     'Предмет':'d_type',
#     'Ссылка':'link',
#     'Работа':'d_nm'
# },inplace=True)
# df[['dtime','d_nm','d_type','link']].to_sql('deadline_gs',con=mcfg_abd.my_connection(),index=False,if_exists='replace')
```
